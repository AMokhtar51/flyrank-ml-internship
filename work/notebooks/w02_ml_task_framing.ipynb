## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple rule can't capture all of the features and separate declining page from non-declining.
not a single feature catches all declining pages unlike a ML model which makes use of all features and finds complex patterns unlike an if-else statement that would make code alot more complex.
For example, `days_since_last_update >= 180 AND impressions_90d >= 500` only flags 17 pages out of 30,000 but 16,262 are actually declining.
That's exactly the kind of pattern a model can pick up by weighing several weak features together, where a fixed rule can only reasonably check one or two thresholds before becoming unmanageable.
