## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Decision:** Which pages should a content reviewer look at first, out of many, given
limited time to review them all?

**Who acts, and how:** a content reviewer works down a ranked queue, spending limited time
on the highest-priority pages first — deciding to refresh, expand, protect, prune, or
monitor each one.

**Cost of a wrong call:**
- False positive: reviewer spends time on a page that didn't need it, while a page that
  did need attention waits longer.
- False negative: a real decline or opportunity is ranked low or missed, and keeps losing
  visibility silently until someone notices by chance.

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()
print(f"Stale but visible candidates: {stale_visible} ({stale_visible/len(df):.1%} of {len(df)} pages)")

declining_with_demand = ((df["trend_direction"] == "down") & (df["impressions_90d"] >= 100)).sum()
print(f"Declining with demand candidates: {declining_with_demand} ({declining_with_demand/len(df):.1%} of pages)")

print(f"Unique clients: {df["client_id"].nunique()}")
```
